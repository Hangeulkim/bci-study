```python
"""
1. STDP `tau_plus`, `tau_minus`를 5/20/50 ms로 바꿔 윈도우 모양 비교.
2. anti-STDP(부호 반전)로 바꾸면 가중치가 어떻게 발산/수렴하는지 관찰.
"""

import numpy as np
import matplotlib.pyplot as plt

A_plus, A_minus = 0.1, 0.12
tau_plus, tau_minus = 20.0, 20.0

dt = np.linspace(-50, 50, 200)

# TODO: STDP 함수
#   dt>0 → A_plus * exp(-dt/tau_plus)
#   dt<0 → -A_minus * exp(dt/tau_minus)
def stdp(tp, tm):
    return np.where(dt > 0, A_plus * np.exp(-dt/tp), -A_minus * np.exp(dt/tm))
dw = stdp(tau_plus, tau_minus)

plt.axhline(0, color='gray'); plt.axvline(0, color='gray')
plt.plot(dt, dw); plt.xlabel("t_post - t_pre (ms)"); plt.ylabel("Δw"); plt.show()
```

```python
# 1. STDP `tau_plus`, `tau_minus`를 5/20/50 ms로 바꿔 윈도우 모양 비교.
tau_list = [5., 20., 50.]

for tau in tau_list:
    dw = stdp(tau, tau)
    
    plt.axhline(0, color='gray'); plt.axvline(0, color='gray')
    plt.plot(dt, dw); plt.xlabel(f"t_post - t_pre (ms) {tau}"); plt.ylabel("Δw"); plt.show()
```

```python
# 2. anti-STDP(부호 반전)로 바꾸면 가중치가 어떻게 발산/수렴하는지 관찰.
tau_list = [-5., -20., -50.]

for tau in tau_list:
    dw = stdp(tau, tau)
    
    plt.axhline(0, color='gray'); plt.axvline(0, color='gray')
    plt.plot(dt, dw); plt.xlabel(f"t_post - t_pre (ms) {tau}"); plt.ylabel("Δw"); plt.show()
```
